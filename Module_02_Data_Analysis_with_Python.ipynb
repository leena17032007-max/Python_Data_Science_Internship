```python
import pandas as pd
import numpy as np

print("=" * 55)
print("MODULE 02 - DATA ANALYSIS WITH PYTHON")
print("Using Pandas and NumPy")
print("=" * 55)

data = {
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Maths": [85, 90, 78, 65, 95, 80],
    "Science": [88, 92, 80, 70, 96, 85],
    "English": [82, 85, 75, 68, 90, 78],
    "Attendance": [95, 98, 85, 75, 99, 88]
}

df = pd.DataFrame(data)

print("\n1. ORIGINAL DATASET")
print(df)

print("\n2. DATASET INFORMATION")
print(df.info())

print("\nDATASET SHAPE:")
print(df.shape)

print("\nCOLUMN NAMES:")
print(df.columns.tolist())

print("\nSTATISTICAL SUMMARY:")
print(df.describe())

df.loc[2, "Maths"] = np.nan
df.loc[3, "Science"] = np.nan

print("\n3. MISSING VALUES BEFORE CLEANING")
print(df.isnull().sum())

df["Maths"] = df["Maths"].fillna(df["Maths"].mean())
df["Science"] = df["Science"].fillna(df["Science"].mean())

print("\nMISSING VALUES AFTER CLEANING")
print(df.isnull().sum())

df = df.drop_duplicates()

print("\n4. DATASET AFTER REMOVING DUPLICATES")
print(df)

df["Total"] = df[["Maths", "Science", "English"]].sum(axis=1)

df["Average"] = df[["Maths", "Science", "English"]].mean(axis=1)

df["Percentage"] = df["Average"]

print("\n5. STUDENT PERFORMANCE")
print(df)

def calculate_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    else:
        return "D"

df["Grade"] = df["Percentage"].apply(calculate_grade)

print("\n6. STUDENT GRADES")
print(df[["Student", "Average", "Grade"]])

top_student = df.loc[df["Average"].idxmax()]

print("\n7. TOP PERFORMER")
print("Student:", top_student["Student"])
print("Average:", round(top_student["Average"], 2))
print("Grade:", top_student["Grade"])

print("\n8. SUBJECT-WISE AVERAGE")
print(df[["Maths", "Science", "English"]].mean())

print("\n9. ATTENDANCE ANALYSIS")
print("Average Attendance:", round(df["Attendance"].mean(), 2), "%")

low_attendance = df[df["Attendance"] < 80]

print("\nStudents with attendance below 80%:")
print(low_attendance[["Student", "Attendance"]])

marks = np.array([
    [85, 88, 82],
    [90, 92, 85],
    [78, 80, 75],
    [65, 70, 68],
    [95, 96, 90],
    [80, 85, 78]
])

print("\n10. NUMPY ANALYSIS")
print("Marks Array:")
print(marks)

print("Overall Mean:", np.mean(marks))
print("Highest Mark:", np.max(marks))
print("Lowest Mark:", np.min(marks))

df.to_csv("student_performance_analysis.csv", index=False)

print("\n11. DATASET SAVED SUCCESSFULLY!")

print("\n" + "=" * 55)
print("MODULE 02 COMPLETED SUCCESSFULLY")
print("Data Analysis using Pandas and NumPy")
print("=" * 55)

display(df)
```
