```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("MODULE 03 - DATA VISUALIZATION")
print("Libraries imported successfully!")

data = {
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Maths": [85, 90, 78, 65, 95, 72],
    "Science": [88, 92, 80, 70, 96, 75],
    "English": [82, 85, 75, 68, 90, 78]
}

df = pd.DataFrame(data)

df["Average"] = df[["Maths", "Science", "English"]].mean(axis=1)

print("\nSTUDENT PERFORMANCE DATASET")
display(df)

plt.figure(figsize=(8, 5))
plt.bar(df["Student"], df["Average"])
plt.title("Student Average Marks")
plt.xlabel("Students")
plt.ylabel("Average Marks")
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(df["Student"], df["Maths"], marker="o", label="Maths")
plt.plot(df["Student"], df["Science"], marker="o", label="Science")
plt.plot(df["Student"], df["English"], marker="o", label="English")
plt.title("Subject-wise Student Performance")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.legend()
plt.grid(True)
plt.show()

subject_avg = df[["Maths", "Science", "English"]].mean()

plt.figure(figsize=(7, 7))
plt.pie(
    subject_avg,
    labels=subject_avg.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Subject-wise Average Marks Distribution")
plt.show()

plt.figure(figsize=(8, 5))
plt.hist(df["Average"], bins=5, edgecolor="black")
plt.title("Distribution of Student Average Marks")
plt.xlabel("Average Marks")
plt.ylabel("Number of Students")
plt.show()

plt.figure(figsize=(8, 5))
plt.scatter(df["Maths"], df["Science"])
plt.title("Maths vs Science Performance")
plt.xlabel("Maths Marks")
plt.ylabel("Science Marks")
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="Student", y="Average")
plt.title("Average Marks by Student")
plt.show()

plt.figure(figsize=(7, 5))
correlation = df[["Maths", "Science", "English"]].corr()
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Subject Marks Correlation Heatmap")
plt.show()

print("\nDATA ANALYSIS AND INSIGHTS")

print("\nSubject-wise Average Marks:")
print(df[["Maths", "Science", "English"]].mean())

top_student = df.loc[df["Average"].idxmax()]
print("\nTop Performing Student:")
print(top_student["Student"])
print("Average Marks:", round(top_student["Average"], 2))

print("\nOverall Average Marks:")
print(round(df["Average"].mean(), 2))

print("\nStudents Scoring Below 75 Average:")
display(df[df["Average"] < 75])

df.to_csv("student_visualization_data.csv", index=False)

print("\nDataset saved successfully!")
print("MODULE 03 COMPLETED SUCCESSFULLY!")
```
