```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("STUDENT PERFORMANCE ANALYSIS")
print("=" * 55)

df = pd.read_csv("/content/student-mat.csv", sep=";")

print("\nDataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nFIRST FIVE RECORDS")
display(df.head())

print("\nDATASET INFORMATION")
df.info()

print("\nSTATISTICAL SUMMARY")
display(df.describe())

print("\nCOLUMN NAMES")
print(df.columns.tolist())

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nTOTAL MISSING VALUES:", df.isnull().sum().sum())

print("\nDATA CLEANING")

print("Duplicate records:", df.duplicated().sum())

df = df.drop_duplicates()

numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    if df[column].isnull().any():
        df[column] = df[column].fillna(df[column].mode()[0])

print("Data cleaning completed!")
print("Remaining missing values:", df.isnull().sum().sum())

print("\nSTUDENT PERFORMANCE ANALYSIS")

df["Average"] = df[["G1", "G2", "G3"]].mean(axis=1)

df["Total"] = df[["G1", "G2", "G3"]].sum(axis=1)

df["Result"] = np.where(df["G3"] >= 10, "Pass", "Fail")

def calculate_grade(mark):
    if mark >= 16:
        return "A"
    elif mark >= 14:
        return "B"
    elif mark >= 12:
        return "C"
    elif mark >= 10:
        return "D"
    else:
        return "F"

df["Grade"] = df["G3"].apply(calculate_grade)

display(df[["G1", "G2", "G3", "Average", "Total", "Result", "Grade"]].head(10))

print("\nOVERALL PERFORMANCE")

print("Average final grade:", round(df["G3"].mean(), 2))
print("Highest final grade:", df["G3"].max())
print("Lowest final grade:", df["G3"].min())

print("\nPASS AND FAIL COUNT")
print(df["Result"].value_counts())

print("\nGRADE DISTRIBUTION")
print(df["Grade"].value_counts())

print("\nTOP 10 PERFORMING STUDENTS")

top_students = df.sort_values(by="G3", ascending=False).head(10)

display(top_students[["school", "sex", "age", "G1", "G2", "G3", "Average"]])

print("\nAVERAGE FINAL GRADE BASED ON STUDY TIME")

study_analysis = df.groupby("studytime")["G3"].mean()

display(study_analysis)

print("\nABSENCE ANALYSIS")

print("Average absences:", round(df["absences"].mean(), 2))
print("Maximum absences:", df["absences"].max())

print("\nAverage final grade by absences")

absence_analysis = df.groupby("absences")["G3"].mean()

display(absence_analysis.head(10))

plt.figure(figsize=(8, 5))
sns.histplot(df["G3"], bins=15, kde=True)
plt.title("Distribution of Final Student Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.show()

plt.figure(figsize=(6, 5))
sns.countplot(data=df, x="Result")
plt.title("Student Pass and Fail Analysis")
plt.xlabel("Result")
plt.ylabel("Number of Students")
plt.show()

plt.figure(figsize=(8, 5))
sns.barplot(x=study_analysis.index, y=study_analysis.values)
plt.title("Average Final Grade by Study Time")
plt.xlabel("Study Time Category")
plt.ylabel("Average Final Grade")
plt.show()

plt.figure(figsize=(8, 5))
df[["G1", "G2", "G3"]].mean().plot(kind="bar")
plt.title("Comparison of Period Grades")
plt.xlabel("Exam Period")
plt.ylabel("Average Grade")
plt.show()

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="absences", y="G3")
plt.title("Student Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade")
plt.show()

plt.figure(figsize=(10, 7))
correlation = df[["age", "studytime", "failures",
                  "absences", "G1", "G2", "G3"]].corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Student Performance Correlation Heatmap")
plt.show()

print("\nPROJECT FINDINGS")
print("=" * 55)

print("Total students:", len(df))
print("Average final grade:", round(df["G3"].mean(), 2))
print("Number of passed students:", (df["G3"] >= 10).sum())
print("Number of failed students:", (df["G3"] < 10).sum())

print("Highest final grade:", df["G3"].max())
print("Lowest final grade:", df["G3"].min())

print("\nStudent Performance Analysis Completed Successfully!")

df.to_csv("student_performance_analysis_cleaned.csv", index=False)

print("\nCleaned dataset saved successfully!")
```
