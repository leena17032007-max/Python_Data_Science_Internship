```python
print("=" * 50)
print("PYTHON & DATA SCIENCE FUNDAMENTALS")
print("Internship Module 01")
print("Name: Leena S")
print("=" * 50)

print("\n1. VARIABLES AND DATA TYPES")

name = "Leena S"
age = 20
department = "Computer Science"
marks = 85.5
is_student = True

print("Name:", name)
print("Age:", age)
print("Department:", department)
print("Marks:", marks)
print("Student:", is_student)

print("\n2. CONDITIONAL STATEMENTS")

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 50:
    grade = "C"
else:
    grade = "Fail"

print("Grade:", grade)

print("\n3. LOOPS")

print("Numbers from 1 to 10:")
for i in range(1, 11):
    print(i, end=" ")

print("\n")

count = 1
while count <= 5:
    print("Count:", count)
    count += 1

print("\n4. LISTS")

student_marks = [85, 90, 78, 88, 92]

print("Student Marks:", student_marks)
print("Highest Marks:", max(student_marks))
print("Lowest Marks:", min(student_marks))
print("Average Marks:", sum(student_marks) / len(student_marks))

print("\n5. DICTIONARIES")

student = {
    "Name": "Leena S",
    "Age": 20,
    "Department": "Computer Science",
    "Marks": 85.5
}

for key, value in student.items():
    print(key, ":", value)

print("\n6. FUNCTIONS")

def calculate_average(marks_list):
    return sum(marks_list) / len(marks_list)

average = calculate_average(student_marks)
print("Average Student Marks:", average)

print("\n7. FILE HANDLING")

with open("student_data.txt", "w") as file:
    file.write("Student Performance Analysis\n")
    file.write("Name: Leena S\n")
    file.write("Department: Computer Science\n")
    file.write("Marks: 85.5\n")

with open("student_data.txt", "r") as file:
    print(file.read())

print("\n8. BASIC DATA SCIENCE ANALYSIS")

import statistics

data = [85, 90, 78, 88, 92, 75, 95]

print("Dataset:", data)
print("Mean:", statistics.mean(data))
print("Median:", statistics.median(data))
print("Mode:", statistics.multimode(data))



print("\n" + "=" * 50)
print("MODULE 01 COMPLETED")
print("Python Fundamentals Practice Successfully Executed")
print("=" * 50)
```
