```python
import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import joblib

sns.set_theme(style="whitegrid")

print("STUDENT PERFORMANCE PREDICTION SYSTEM")
print("=" * 55)

file_path = "/content/student-mat.csv"

if not os.path.exists(file_path):
    raise FileNotFoundError(
        "student-mat.csv not found. Upload the dataset using the Files panel."
    )

df = pd.read_csv(file_path, sep=";")

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nFirst five records:")
display(df.head())

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe())

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate records:", df.duplicated().sum())

df = df.drop_duplicates()

df = df.dropna(subset=["G3"])

print("\nDataset after cleaning:", df.shape)

df["Result"] = np.where(df["G3"] >= 10, "Pass", "Fail")

print("\nPass and Fail counts:")
print(df["Result"].value_counts())

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Result")
plt.title("Student Pass and Fail Distribution")
plt.xlabel("Result")
plt.ylabel("Number of Students")
plt.show()

plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="G3", bins=15, kde=True)
plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")
plt.show()

plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="studytime", y="G3")
plt.title("Study Time vs Final Grade")
plt.xlabel("Study Time")
plt.ylabel("Average Final Grade")
plt.show()

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="absences", y="G3")
plt.title("Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade")
plt.show()

plt.figure(figsize=(12, 8))
numeric_df = df.select_dtypes(include=np.number)
sns.heatmap(numeric_df.corr(), cmap="coolwarm", annot=False)
plt.title("Student Dataset Correlation Heatmap")
plt.show()

X = df.drop(columns=["G1", "G2", "G3", "Result"])
y = df["Result"]

numeric_features = X.select_dtypes(include=np.number).columns.tolist()
categorical_features = X.select_dtypes(exclude=np.number).columns.tolist()

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))

model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight="balanced"
    ))
])

model.fit(X_train, y_train)

print("\nMachine Learning model trained successfully!")

y_pred = model.predict(X_test)

print("\nSample Predictions:")
results = pd.DataFrame({
    "Actual Result": y_test.values,
    "Predicted Result": y_pred
})
display(results.head(10))

accuracy = accuracy_score(y_test, y_pred)

print("\nMODEL PERFORMANCE")
print("=" * 40)
print(f"Accuracy: {accuracy * 100:.2f}%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

cm = confusion_matrix(y_test, y_pred, labels=["Fail", "Pass"])

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Fail", "Pass"]
)

disp.plot(cmap="Blues")
plt.title("Student Result Prediction - Confusion Matrix")
plt.show()

new_student = X.iloc[[0]].copy()

new_prediction = model.predict(new_student)[0]

print("\nNEW STUDENT PREDICTION")
print("=" * 40)
print("Predicted Result:", new_prediction)

df.to_csv(
    "/content/student_performance_cleaned.csv",
    index=False
)

results.to_csv(
    "/content/student_prediction_results.csv",
    index=False
)

joblib.dump(
    model,
    "/content/student_performance_model.pkl"
)

print("\nPROJECT FILES SAVED SUCCESSFULLY!")
print("1. student_performance_cleaned.csv")
print("2. student_prediction_results.csv")
print("3. student_performance_model.pkl")

print("\n" + "=" * 55)
print("PROJECT COMPLETED!")
print("=" * 55)
print("Project: Student Performance Prediction System")
print("Dataset: UCI Student Performance - Mathematics")
print("Tools: Python, Pandas, NumPy, Matplotlib, Seaborn")
print("Machine Learning: Random Forest Classifier")
print(f"Model Accuracy: {accuracy * 100:.2f}%")
print("Target: Predict Pass or Fail")
print("=" * 55)
```
